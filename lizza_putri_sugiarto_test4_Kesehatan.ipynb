```python
import yt_dlp
import pandas as pd
from datetime import datetime, timezone
import time

# Masukkan URL video dengan tema kesehatan jantung
video_urls = [
    "https://youtu.be/3P24C_EMrlU",
    "https://youtu.be/tFn7dgJuyDE?si=okraumgJE78Xh4US",
    "https://youtu.be/We3dl8c8xw4?si=ksOqUQjxbbVw0b-6",
    "https://youtu.be/rYIty04HVHk?si=P0ESlPxckLkfOUgc",
]

all_comments = []

for url in video_urls:

    if "MASUKKAN_URL" in url:
        continue

    print("\nMengambil:", url)

    options = {
        "quiet": True,
        "skip_download": True,
        "getcomments": True,
        "extractor_args": {
            "youtube": {
                "max_comments": ["300,300,0,0,1"]
            }
        }
    }

    try:
        with yt_dlp.YoutubeDL(options) as ydl:
            info = ydl.extract_info(url, download=False)

        comments = info.get("comments") or []

        for c in comments:
            all_comments.append({
                "video_id": info.get("id"),
                "video_title": info.get("title"),
                "created_at": c.get("timestamp"),
                "username": c.get("author"),
                "full_text": c.get("text")
            })

        print("Komentar diperoleh:", len(comments))

    except Exception as e:
        print("Gagal:", e)

    time.sleep(2)

# Gabungkan seluruh data
df_all = pd.DataFrame(all_comments)

# Hapus komentar kosong
df_all = df_all.dropna(subset=["full_text"])

# Bersihkan spasi
df_all["full_text"] = df_all["full_text"].astype(str).str.strip()

df_all = df_all[df_all["full_text"] != ""]

# Hapus duplikat
df_all = df_all.drop_duplicates(subset=["full_text"])

# Konversi timestamp
df_all["created_at"] = pd.to_datetime(
    df_all["created_at"],
    unit="s",
    utc=True
).dt.tz_convert("Asia/Jakarta")

# Simpan dataset mentah gabungan
df_all.to_csv(
    "youtube_kesehatan_raw.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nTOTAL DATA UNIK:", len(df_all))
print("JUMLAH VIDEO:", df_all["video_id"].nunique())

display(df_all.head(10))
```

```python
import pandas as pd
import re

# Membuat salinan dataset mentah
df_clean = df_all.copy()

# Fungsi pembersihan teks
def clean_text(text):
    text = str(text).lower()

    # Menghapus URL
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Menghapus mention
    text = re.sub(r'@\w+', ' ', text)

    # Menghapus hashtag tanpa menghilangkan kata setelahnya
    text = re.sub(r'#', '', text)

    # Menghapus angka
    text = re.sub(r'\d+', ' ', text)

    # Menghapus simbol yang tidak relevan
    text = re.sub(r'[^a-zA-ZÀ-ÿ\s]', ' ', text)

    # Menghapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Terapkan preprocessing
df_clean["clean_text"] = df_clean["full_text"].apply(clean_text)

# Menghapus hasil kosong
df_clean = df_clean.dropna(subset=["clean_text"])
df_clean = df_clean[df_clean["clean_text"] != ""]

print("Preprocessing selesai!")
print("Jumlah data awal:", len(df_all))
print("Jumlah data setelah cleansing:", len(df_clean))

display(df_clean[["full_text", "clean_text"]].head(15))
```

```python
# Kamus kata positif
positive_words = [
    "bagus", "baik", "sehat", "sembuh", "pulih",
    "bermanfaat", "membantu", "menginspirasi",
    "terima kasih", "makasih", "suka", "senang",
    "setuju", "mantap", "luar biasa", "hebat",
    "jelas", "mudah", "penting", "aman",
    "beruntung", "bersyukur", "alhamdulillah",
    "semangat", "sukses", "puas", "terbaik",
    "edukatif", "menarik", "terbantu", "mengerti"
]

# Kamus kata negatif
negative_words = [
    "buruk", "bahaya", "sakit", "nyeri",
    "sesak", "serangan", "meninggal", "mati",
    "parah", "takut", "khawatir", "cemas",
    "sedih", "kecewa", "gagal", "susah",
    "sulit", "lemah", "darurat", "risiko",
    "resiko", "penyakit", "gangguan", "pusing",
    "kolesterol", "hipertensi", "aritmia",
    "palpitasi", "jantung koroner", "stroke",
    "merokok", "rokok", "obesitas", "komplikasi",
    "terlambat", "memburuk", "tidak nyaman"
]

print("Jumlah kata positif:", len(positive_words))
print("Jumlah kata negatif:", len(negative_words))
```

```python
import pandas as pd

def label_sentiment(text):

    text = str(text).lower()

    positive_count = sum(
        text.count(word) for word in positive_words
    )

    negative_count = sum(
        text.count(word) for word in negative_words
    )

    if positive_count > negative_count:
        sentiment = "positif"

    elif negative_count > positive_count:
        sentiment = "negatif"

    else:
        sentiment = "netral"

    return pd.Series({
        "positive_score": positive_count,
        "negative_score": negative_count,
        "sentimen": sentiment
    })


# Terapkan pelabelan ke seluruh dataset
hasil_label = df_clean["clean_text"].apply(label_sentiment)

# Gabungkan hasil ke DataFrame
df_clean = pd.concat(
    [df_clean, hasil_label],
    axis=1
)

print("Pelabelan berhasil!")
print("Kolom dataset:", df_clean.columns.tolist())

display(df_clean[
    ["clean_text", "positive_score",
     "negative_score", "sentimen"]
].head(10))
```

```python
import re
import pandas as pd

# Kata negasi
negation_words = {
    "tidak", "bukan", "belum", "jangan",
    "tanpa", "kurang", "gak", "enggak",
    "nggak", "ga"
}

def count_words_with_negation(text, word_list):

    words = text.split()
    score = 0

    for phrase in word_list:

        phrase_tokens = phrase.lower().split()
        phrase_len = len(phrase_tokens)

        for i in range(len(words) - phrase_len + 1):

            candidate = words[i:i + phrase_len]

            if candidate == phrase_tokens:

                # Periksa dua kata sebelum kata sentimen
                start = max(0, i - 2)
                previous_words = words[start:i]

                has_negation = any(
                    word in negation_words
                    for word in previous_words
                )

                if has_negation:
                    score -= 1
                else:
                    score += 1

    return score


def improved_label_sentiment(text):

    text = str(text).lower()

    positive_score = count_words_with_negation(
        text, positive_words
    )

    negative_score = count_words_with_negation(
        text, negative_words
    )

    # Penentuan label berdasarkan skor
    if positive_score > negative_score:
        sentiment = "positif"

    elif negative_score > positive_score:
        sentiment = "negatif"

    else:
        sentiment = "netral"

    return pd.Series({
        "positive_score": positive_score,
        "negative_score": negative_score,
        "sentimen": sentiment
    })


# Terapkan ulang pada dataset bersih
new_labels = df_clean["clean_text"].apply(
    improved_label_sentiment
)

# Ganti kolom label lama
df_clean = df_clean.drop(
    columns=["positive_score", "negative_score", "sentimen"],
    errors="ignore"
)

df_clean = pd.concat(
    [df_clean, new_labels],
    axis=1
)

print("Pelabelan ulang selesai!")
display(
    df_clean[
        ["full_text", "positive_score",
         "negative_score", "sentimen"]
    ].head(15)
)
```

```python
import re

def normalize_words(text):

    text = str(text).lower()

    # Mengurangi huruf berulang yang berlebihan
    # contoh: sukaaa -> suka
    # contoh: sehattt -> sehat
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # Normalisasi beberapa kata informal
    replacements = {
        "aq": "aku",
        "gw": "saya",
        "gue": "saya",
        "gak": "tidak",
        "ga": "tidak",
        "nggak": "tidak",
        "enggak": "tidak",
        "makasih": "terima kasih"
    }

    words = text.split()

    normalized = [
        replacements.get(word, word)
        for word in words
    ]

    return " ".join(normalized)


# Terapkan normalisasi
df_clean["clean_text"] = (
    df_clean["clean_text"].apply(normalize_words)
)

display(
    df_clean[["full_text", "clean_text"]].head(15)
)
```

```python
# Menghitung ulang label dari clean_text terbaru
new_labels = df_clean["clean_text"].apply(
    improved_label_sentiment
)

# Menghapus kolom label sebelumnya
df_clean = df_clean.drop(
    columns=[
        "positive_score",
        "negative_score",
        "sentimen"
    ],
    errors="ignore"
)

# Memasukkan hasil label terbaru
df_clean = pd.concat(
    [df_clean, new_labels],
    axis=1
)

print("Pelabelan ulang berhasil!")
```

```python
import re
import pandas as pd

negation_words = {
    "tidak", "bukan", "belum", "jangan",
    "tanpa", "gak", "ga", "enggak", "nggak"
}

def improved_sentiment_v2(text):

    words = str(text).lower().split()

    positive_score = 0
    negative_score = 0

    def is_negated(index):
        previous = words[max(0, index-2):index]
        return any(w in negation_words for w in previous)

    for word in positive_words:

        phrase = word.lower().split()
        n = len(phrase)

        for i in range(len(words)-n+1):

            if words[i:i+n] == phrase:

                if is_negated(i):
                    negative_score += 1
                else:
                    positive_score += 1

    for word in negative_words:

        phrase = word.lower().split()
        n = len(phrase)

        for i in range(len(words)-n+1):

            if words[i:i+n] == phrase:

                if is_negated(i):
                    positive_score += 1
                else:
                    negative_score += 1

    if positive_score > negative_score:
        label = "positif"

    elif negative_score > positive_score:
        label = "negatif"

    else:
        label = "netral"

    return pd.Series({
        "positive_score": positive_score,
        "negative_score": negative_score,
        "sentimen": label
    })


# Pelabelan ulang
new_labels = df_clean["clean_text"].apply(
    improved_sentiment_v2
)

df_clean = df_clean.drop(
    columns=[
        "positive_score",
        "negative_score",
        "sentimen"
    ],
    errors="ignore"
)

df_clean = pd.concat(
    [df_clean, new_labels],
    axis=1
)

print("Pelabelan versi 2 selesai!")
display(df_clean.head(10))
```
